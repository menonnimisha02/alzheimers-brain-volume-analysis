```python
import pandas as pd
```

```python
df = pd.read_excel("oasis_longitudinal_demographics-8d83e569fa2e2d30.xlsx")
```

```python
df.shape
```

```python
df.dtypes
```

```python
df = df.drop(columns=['MRI ID','Hand','ASF'])
```

```python
mapping = {
    "Subject ID" : "patient_id",
    "Group" : "group",
    "Visit" : "visit",
    "MR Delay" : "mr_delay",
    "M/F" : "gender",
    "Age" : "age",
    "EDUC" : "education",
    "SES" : "ses",
    "MMSE" : "mmse",
    "CDR" : "cdr",
    "eTIV" : "etiv",
    "nWBV" : "nwbv"
}

df = df.rename(columns = mapping)
```

```python
df.isna().sum()
```

```python
df.value_counts()
```

```python
df[df['ses'].isna()]
```

```python
df['ses'] = df['ses'].fillna(df['ses'].median())
```

```python
df['ses'] = df['ses'].astype(int)
```

```python
df['mmse'] = df.groupby(['cdr','visit'])['mmse'].transform(lambda x: x.fillna(x.median()))
```

```python
df['dataset_source'] = 'oasis_2'
```
